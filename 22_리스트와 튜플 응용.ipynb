```python
#  리스트의 가장 작은수, 가장 큰수, 합계 구하기
a = [38, 21, 53, 62, 19]
smallest = a[0]
for i in a:
    if i < smallest:
        smallest = i

print(smallest)
```

```python
# 리스트 컴프리헨션
## 1.[식 for 변수 in 리스트]
## 2.list([식 for 변수 in 리스트])

a = [i for i in range(10)]
print(a)

# for i in range(10):
#     print(i)

b = list(j for j in range(10))
print(b)
```

```python
# 리스트 표현식에서 if 조건문 사용하기
## 1.[식 for 변수 in 리스트 if 조건식]
## 2. list(식 for 변수 in 리스트 if 조건식)

a = [i for i in range(10) if i % 2 == 0]
print(a)
```

```python
b = [j + 5 for j in range(10) if j % 2 == 1]
b
```

```python
# for 반복문과 if 조건문을 여러 번 사용하기
# [식 for 변수1 in 리스트1 if 조건식1
#    for 변수2 in 리스트2 if 조건식2]

a = [ 
    i * j for j in range(2,10) 
     for i in range(1, 10)
     ]
print(a)
```

```python
# 리스트에 맵 사용하기

## 1. list(map(함수, 리스트))
## 2. tuple(map(함수, 튜플))

a = [1.2, 2.5, 3.7, 4.6]
for i in range(len(a)):
    a[i] = int(a[i])

print(a)
```

```python
start, end = list(map(int,input().split()))
if start<1 or start>20 or end<10 or end>30 or start>end:
    print("주어진 범위를 확인하세요")
    
else:
    a = []
    for i in range(start, end + 1):
        if i == start + 1 or i == end - 1:
            continue
        else:
            a.append(2 ** i)

    print(a)
```

```python
height = int(input("높이를 입력하세요: "))

for i in range(height, 0, -1):

    for j in range(height - i):
        print(" ", end='')

    for j in range(2 * i - 1):
        print("*", end='')

    print()
```
