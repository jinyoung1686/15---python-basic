```python
# format 메서드로 문자열 정렬하기 -> 콜론붙인후 정렬할 방향, 길이 지정
print('{0:10}'.format('python')) # 기본은 왼쪽정렬
print('{0:<10}'.format('python'))
print('{0:>10}'.format('python'))
print('{:>10}'.format('python')) # 인덱스 사용하지 않을때
```
