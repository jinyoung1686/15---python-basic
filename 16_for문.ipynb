```python
# for 와 range 사용하기
# for 변수 in range:
#     반복할코드

for i in range(100):
    print("hello world")
```

```python
# 반복문에 변수의 변화 알아보기
for i in range(100):
    print("hello world", i)
```

```python
for i in range(5, 12):
    print("hello world",i)
```

```python
for imsi in range(0, 10 ,2):
     print("hello world",imsi)
```

```python
for a in range(10, 0, -1):
     print("hello world", a)


list(range(10,0))
```

```python
print(list(range(10)))
print(list(reversed(range(10))))
```

```python
for c in reversed(range(10)):
     print("hello world",c)
```

```python
# 시퀀스 객체로 반복하기

a = [10, 20, 30, 40, 50]
for imsi in a:
    print(imsi)
```

```python
fruits = ('apple', 'orange','grape')
for fruit in fruits:
    print(fruit)
```

```python
for letter in 'Python':
    print(letter,end=' ')
```

```python
dan = int(input("단을 입력하세요: "))

for i in range(1, 10):
    print(f"{dan} x {i} = {dan * i}")
```
