```python
#값을 여러개 출력하기

print(1, 2, 3)
```

```python
#sep로 값 사이에 문자넣기 
print(1, 2, 3, sep=', ')
```

```python
print(1920, 1000,sep="x")
```

```python
# 줄바꿈 활용하기
print(1, 2, 3, sep='\n')
```

```python
print("1\n2\n3")
```

```python
# end 사용하기(기본값은 줄바꿈)
print(1)
print(2)
print(3)
```

```python
print(1, end="")
print(2, end="")
print(3)
```

```python
print(1, end=" ")
print(2, end=" ")
print(3)
print(1, 2, 3)
```

```python
print(3.1, 'python', 100)
print(16, 9, sep=":")
print('Hello\npython')
print('hello','\n','python',sep='')

year= 2000
month = 10
day = 27
hour = 11
minute = 43
second = 59

print(year, month, day, sep='/',end='')
print(hour,minute,second,sep=':')
# print에서 end 기본값은 end="\n", 이고 sep 기본값은 sep=" " 임
```
