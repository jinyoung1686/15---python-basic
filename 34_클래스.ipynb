```python
# 인스턴스를 만들때 값 받기
# class 클래스이름:
#     def __init__(self, 매개변수1, 매개변수2):
#         self.속성1 = 매개변수1
#         self.속성2 = 매개변수2

class Person:
    def __init__(self, name, age, address):
        self.hello = '안녕하세요.'
        self.name = name
        self.age = age
        self.address = address

    def greeting(self):
        print(f'{self.hello} 저는 {self.name}입니다.')

maria = Person('마리아', 20, '서울시 서초구 반포동')
maria.greeting()

print('이름:' , maria.name)
print('나이:' , maria.age)
print('주소:' , maria.address)
```

```python
# 비공개 속성 사용하기
# class 클래스이름:
#     def __init__(self, 매개변수1):
#         self.속성1 = 매개변수1
#         self.__속성2 = 값

class Person:
    def __init__(self, name, age, address,wallet):
        self.hello = '안녕하세요.'
        self.name = name
        self.age = age
        self.address = address
        self.__wallet = wallet # -> 비공개 속성(__)

maria = Person('마리아', 20, '서울시 서초구 반포동', 10000)
maria.__wallet -= 10000
```

```python
class Person:
    def __init__(self, name, age, address,wallet):
        self.hello = '안녕하세요.'
        self.name = name
        self.age = age
        self.address = address
        self.__wallet = wallet # -> 비공개 속성(__)


    def pay(self, amount):
        self.__wallet -= amount
        print(f'이제{self.__wallet}원 남았어요.')



maria = Person('마리아', 20, '서울시 서초구 반포동', 10000)
maria.pay(3000)
```

```python
class Person:
    def __init__(self, name, age, address,wallet):
        self.hello = '안녕하세요.'
        self.name = name
        self.age = age
        self.address = address
        self.__wallet = wallet # -> 비공개 속성(__)


    def pay(self, amount):
        if amount > self.__wallet:
            print('돈이 없어..')
            return
        self.__wallet -= amount
        print(f'이제{self.__wallet}원 남았어요.')



maria = Person('마리아', 20, '서울시 서초구 반포동', 10000)
maria.pay(5000)

```

```python
health, mana, ability_power = map(float,(input().split()))

class Annie:
    def __init__(self, health, mana, ability_power):
        self.health = health
        self.mana = mana
        self.ability_power = ability_power

    def tibbers(self):
        print(f" 티버: 피해량 {self.ability_power * 0.65 + 400}")

x = Annie(health=health, mana= mana, ability_power = ability_power)
x.tibbers()
```

```python

posts = [
    {"id": 1, "title": "파이썬 공부", "writer": "김철수"},
    {"id": 2, "title": "오늘 할 일", "writer": "이영희"},
    {"id": 3, "title": "장보기 목록", "writer": "박민수"},
    {"id": 4, "title": "파이썬 복습", "writer": "김철수"},
    {"id": 5, "title": "운동 계획", "writer": "최지수"}
]

num = int(input("삭제할 번호: "))

posts = list(filter(lambda x: x["id"] != num, posts))

for x in posts:
    print(*x.values()) 
```
