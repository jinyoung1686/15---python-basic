```python
# elif 사용하기
x = 20
if x == 10:
    print("10입니다")
elif x == 20:
    print("20입니다")
```

```python
# 음료수 자판기

button = int(input())

if button == 1:
    print("콜라")
elif button == 2:
    print("사이다")
elif button == 3:
    print("환타")
else:
    print("제공하지 않는 메뉴")
```

```python
age = int(input())
balance = 9000

if age >= 7 and age <= 12:
    balance -= 650
elif age >= 13 and age <= 18:
    balance -= 1050
elif age >= 19:
    balance -= 1250
else: 
    print('나이를 다시입력해주세요')

print(balance)

```
