```python
# grobal 전역변수
x = 10
def foo():
    global x # x를 전역 변수 만듦
    x = 20
    print(x)
foo()
print(x)
```

```python
# nonlocal 지역ㅂㄴ수
def A():
    x = 10
    def B():
        nonlocal x
        x = 20
    B()
    print(x)

A()
```

```python
# nonlocal이 변수를 찾는 순서

def A():
    x = 10
    y = 100
    def B():
        x = 20
        def C():
            nonlocal x
            nonlocal y
            x = x + 30
            y = y + 300
            print(x)
            print(y)
    
        C()

    B()

A()    

```

```python
# global로 전역 변수 사용하기
x = 1
def A():
    x = 10
    def B():
        x = 20
        def C():
            global x
            x = x + 30
            print('1', x)
        C()
        print('2', x)
    B()
    print('3', x)

A()
print('4', x)
```

```python
# 클로저 사용하기
def calc():
    a = 3
    b = 5
    def mul_add(x):
        return a * x + b
    return mul_add

c = calc()
print(c(1), c(2), c(3), c(4), c(5))
```

```python
# lambda로 클로저 만들기

def calc():
    a = 3
    b = 5
    return lambda x: a * x + b

c = calc()
print(c(1), c(2), c(3), c(4), c(5))
```

```python
# 클로저의 지역 변수 변경하기
def calc():
    a = 3
    b = 5
    total = 0
    def mul_add(x):
        nonlocal total
        total = total + a * x + b
        print(total)
    return mul_add
c = calc()
c(1)
c(2)
c(3)
```
