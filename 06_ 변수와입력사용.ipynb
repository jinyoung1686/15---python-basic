```python
# 입력 값을 정수로 변환하기
a =int(input("첫번째 숫자: "))
b =int(input("두번째 숫자: "))
print(a + b)
```

```python
# 입력 값을 변수 두 개에 저장하기
a, b = input("문자열 2개를 입력하세요: ").split() # 'hello python',split()
print(a)
print(b)
```

```python
# "hello python.split()
a, b = ['hello', 'python']
print(a)
print(b)
```

```python
# 두 숫자의 합 구하기
a, b = input("숫자 2개를 입력하세요: ").split()
a = int(a)
b = int(b)
print(a)
print(b)
print(a + b)
```

```python
# 입력받은 값을 공백를 기준으로 분리하기
a ,b = map(int, input("숫자 두 개를 입력하세요: ").split())

# a ,b = map(int, ["10", "20"])

# a, b= 10, 20

print(a + b)
```

```python
# 입력받은 값을 콤마를 기준으로 분리하기
a ,b = map(int, input("숫자 두 개를 입력하세요: ").split(','))

print(a + b)
```
