```python
# 문자열 판단하기
# re.match('패턴','문자열')

import re

re.match('Hello', 'Hello, world!')
re.match('Python', 'Hello, world!')
```

```python
# 문자열이 맨앞에 오는지 맨뒤에 오는지 판단하기
# re.search('패턴','문자열')

re.search('^Hello', 'Hello, world!')
```

```python
re.search('world!$', 'Hello, world!')
```

```python
# 범위 판단하기
# [0-9]* 숫자 0~9까지 0개이상이면 매칭
# [0-9]+ 숫자 0~9까지 1개이상이면 매칭

re.match('[0-9]*', '1234')

```

```python
re.match('[0-9]+', '1234')
```

```python
re.match('[0-9]+', 'abcd') # 숫자 1개이상 없으므로 매칭 x
```

```python
re.match('a*b', 'b')
```

```python
re.match('a+b', 'b')
```

```python
re.match('a*b', 'ab')
```

```python
re.match('a+b', 'aab')
```

```python
# 문자? -> 문자가 1개 or 없을때
# [0~9]? -> 0~9 까지 숫자가 1개 or 없을때
# . -> 아무문자 한개
re.match('abc?d', 'abd')
```

```python
re.match('ab[0-9]?c', 'ab3c')
```

```python
re.match('ab.d', 'abxd')
```

```python
# 문자 개수 판단하기
## 문자{개수}
## (문자열){개수}

print(re.match('h{3}', 'hhhello'))
print(re.match('(hello){3}', 'hellohellohelloworld'))
```

```python
## [0~9]{개수}

print(re.match('[0-9]{3}-[0-9]{4}-[0-9]{4}', '010-6327-1686'))
print(re.match('[0-9]{3}-[0-9]{4}-[0-9]{4}', '010-6327-168'))
```

```python
## (문자){시작개수,끝개수} 
## (문자열){시작개수,끝개수} 
## [0-9]{시작개수,끝개수} 

print(re.match('[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}', '02-100-1000'))
```

```python
# a-z
# A-Z
print(re.match('[a-zA-Z0-9]+', 'Hello1234'))
print(re.match('[A-Z0-9]+', 'hello'))
```

```python

print(re.match('[가-힣]+', '홍길동'))
print(re.match('[가-힣]+', 'abc홍길동')) # 처음부터 안맞아서 none
```

```python
# 특정 문자 범위에 포함되지 않는지 판단하기
# [^범위]*
# [^범위]+

print(re.match('[^A-Z]+', 'Hello'))
print(re.match('[^A-Z]+', 'hello'))
```

```python
# ^[범위]*
# ^[범위]+

re.search('[A-Z]+', 'Hello')
```

```python
# [범위]*$
# [범위]+$

re.search('[0-9]+$' , 'Hello1234')
```

```python
# 특수문자 판단하기
# \특수문자
# \d -> [0-9]와 같음
# \D -> [^0-9]와 같음, 숫자를 제외한 모든 숫자
# \w -> [a-zA-Z0-9] 영문 대소문자, 숫자, 밑줄문자
# \W -> [^a-zA-Z0-9_] 영문 대소문자, 숫자, 밑줄문자를 제외한 모든 문자

print(re.match('\\d+', '1234'))
print(re.search('\*+', '1 **2'))
```

```python
re.match('[$()a-zA-Z0-9]+', '$(document)')
```

```python
# 공백 처리하기
# \s -> \t \n \r \f
# \S -> 공백제외

print(re.match('[a-zA-Z0-9]+', 'Hello 1234'))
print(re.match('[a-zA-Z0-9\s]+', 'Hello 1234'))
```

```python
# 그룹 사용하기
# (정규표현식)(정규표현식)
# 매치객체.group(그룹숫자)

m = re.match('([0-9]+) ([0-9]+)', '10 295')
print(m.group(1))
print(m.group(2))
print(m.group(0))
print(m.groups())
```

```python
# (?P<이름>정규표현식)
## 매치객체.group(그룹이름)

import re

m = re.match(r'(?P<func>[a-zA-Z_][a-zA-Z0-9_]+)\((?P<arg>\w+)\)', 'print(1234)')

print(m.group('func'))
print(m.group('arg'))
```

```python
## re.findall('패턴','문자열')
re.findall('[0-9]+', '1 2 Fizz 4 Buzz Fizz 7 8')
```

```python
# 문자열 바꾸기
# re.sub('패턴','바꿀문자열','문자열',바꿀횟수)

re.sub('apple|orange', 'fruit', 'apple box orange tree')
```

```python
re.sub('[0-9]+' , 'n', '1 2 Fizz 4 Buzz Fizz 7 8')
```

```python
# 교체함수(매치객체)
## re.sub('패턴',교체함수,'문자열',바꿀횟수)

def multiple10(m):
    n = int(m.group())
    return str(n * 10)

re.sub('[0-9]+', multiple10, '1 2 Fizz 4 Buzz Fizz 7 8')
```

```python
re.sub('[0-9]+', lambda m: str(int(m.group()) * 10), '1 2 Fizz 4 Buzz Fizz 7 8 ')
```

```python
# 찾은 문자열을 결과에 다시 사용하기
#\\ 숫자
import re

print(re.sub(r'([a-z]+)\s([0-9]+)', r'\2 \1 \2 \1', 'hello 1234'))
```
