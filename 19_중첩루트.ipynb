```python
for i in range(5):
    for j in range(5):
        print("j: ",j ,sep=" ",end=" ")
    
    print('i: ', i, '\\n',sep='')
```

```python
# 사각형으로 별 출력하기

for i in range(5):
    for j in range(5):
        print('*',end='')
    print()
    
```

```python
# 사각형 모양 바꾸기

for i in range(3):
    for j in range(7):
        print('*',end='')
    print()
```

```python
# 계단식으로 별 출력하기

for i in range(5):
    for j in range(5):
        if j <= i:
            print('*',end='')
    print()
```

```python
# 대각선으로 별 출력하기

for i in range(5):
    for j in range(5):
        if j == i:
            print('*', end='')
        else:
            print(' ',end='')
    print()
```

```python
height = int(input("높이: "))

for i in range(1, height + 1):
    print(' ' * (height - i) + '*' * (2 * i - 1))
```

```python
height = int(input("높이 입력: "))

for i in range(1, height + 1):
    
    # 공백 출력
    for j in range(height - i):
        print(' ', end='')

    # 별 출력
    for k in range(2 * i - 1):
        print('*', end='')

    print()
```
