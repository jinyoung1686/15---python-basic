```python
# 딕셔너리 표현식에서 if 조건문 사용하기
x = {
    'a' : 10,
    'b' : 20,
    'c' : 30,
    'd' : 40
    }

for key, value in x.items(): # 반복문이 돌아가는 도중에 딕셔너리 크기가 바뀌어서 에러
    if value == 20:
        del x[key]

print(x)
```

```python
# 딕셔너리 표현식에서 if 조건문 사용하기
x = {
    'a' : 10,
    'b' : 20,
    'c' : 30,
    'd' : 40
    }
y = x.copy()

for key, value in list(y.items()): # # list로 다른메모리로 복제하여 할당된 주소가 다름, 따라서 기존메모리는 그대로
    if value == 20:
        del x[key]

print(x)
```

```python

row, col = map(int, input().split())

matrix = []
for i in range(row):
    matrix.append(list(input()))
    
print()
    
for i in range(row):
    for j in range(col):
        cnt = 0
        if matrix[i][j] == ".":
            for y in range(i-1, i+2):
                for x in range(j-1, j+2):
                    if not (y < 0 or x < 0 or y >= row or x >= col):
                        if matrix[y][x] == "*":
                            cnt += 1
            matrix[i][j] = cnt
            print(matrix[i][j], end="")
        else:
            print(matrix[i][j], end="")
    print()
```

```python
## 지뢰찾기 문제 ##
row, col = map(int, input().split())

matrix = []
for i in range(row):
    matrix.append(list(input()))


for i in range(row):
    for j in range(col):
        count = 0
        if matrix[i][j] == ".":
            for y in range(i-1, i+2):
                for x in range(j-1, j+2):
                    if 0 <= y < row and 0 <= x < col:
                        if matrix[y][x] == "*":
                            count += 1
            print(count, end="")
        else:
            print("*", end="")
    print()
```
