```python
# 위치 인수를 사용하는 함수를 만들고 호출하기
def print_number(a, b, c):
    print(a)
    print(b)
    print(c)

print_number(10, 20,{'진영'})
```

```python
# 언패킹 사용하기
print_number(*[10, 20, 30])
```

```python
print_number(10, 20)
```

```python
print_number(10, 20, 30, 40)
```

```python
# 가변 인수 함수 만들기
## def 함수이름(*매개변수):
##    코드

def print_numbers(*args):
    for arg in args:
        print(arg)

print_numbers(10)  
print_numbers(10, 20, 30, 40)  
```

```python
x = [10]
print_numbers(*x)

y = [10, 20, 30, 40]
print_numbers(*y)
```

```python
def print_numbers(a,*args):
    print(a)
    print(args)

print_numbers(1)
```

```python
print_numbers(1, 10, 20)
```

```python
# 키워드 인수 사용하기

def personal_info(name,age,address):
    print('이름:', name)
    print('나이:', age)
    print('주소:', address)
```

```python
personal_info('홍길동', 30, '서울시 용산구 이촌동')
```

```python
personal_info(name = '홍길동', age=30, address='서울시 용산구 이촌동')
```

```python
personal_info(age=30, address='서울시 용산구 이촌동', name = '홍길동')
```

```python
# 딕셔너리 언패킹 - 함수(**딕셔너리)
x = {'name':'홍길동', 'age': 30, 'address':'서울시 용산구'}
personal_info(**x)
```

```python
# 키워드 인수를 사용하는 가변 인수 함수 만들기
## def 함수이름(**매개변수)
##    코드
def personal_info(**kwargs):
    print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, ': ', arg, sep='')
```

```python


personal_info(name ='홍길동',age=30, address='서울시 용산구')
```

```python
# 매개변수에 초깃값 저장하기
## def 함수이름(매개변수=값):
##    코드

def personal_info(name, age, address='비공개'):
    print('이름:', name)
    print('나이:', age)
    print('주소:', address)
```

```python
korean, english, mathematics, science = map(int,input().split())

def get_min_max_score(*args):
    return min(args), max(args)

def get_average(**kwargs):
    return sum(kwargs.values()) / len(kwargs)

min_score, max_score = get_min_max_score(korean, english, mathematics, science)
average_score = get_average(korean=korean, english=english, mathematics=mathematics, science=science)

print(f'높은 점수: {max_score:.2f}, 낮은 점수: {min_score:.2f}, 평균 점수: {average_score:.2f}')
```

```python
scores = list(map(int,input().split()))

def mx_mi_to_avg_score(*args):
    return min(args), max(args), sum(args), sum(args)/len(args)

def bunsan(*args):
    average = sum(args) / len(args)
    total = 0

    for arg in args:
        total += (arg - average) ** 2
    
    return total / len(args)

minscore, maxscore, totalscore, average = mx_mi_to_avg_score(*scores)

bun = bunsan(*scores)

print(f'높은 점수: {maxscore:.2f}, 낮은 점수: {minscore:.2f}, 총 점수: {totalscore:.2f}, 평균 점수: {average:.2f}, 분산: {bun:.2f}')
```

```python
scores = list(map(int, input().split()))

def mx_mn_to_avg_score(*args):
    return min(args), max(args), sum(args), sum(args) / len(args)

def var(*args):
    average = sum(args) / len(args)
    total = 0

    for arg in args:
        total += (arg - average) ** 2

    return total / len(args)

min_score, max_score, total_score, average_score = mx_mn_to_avg_score(*scores)

var_score = var(*scores)

print(f'높은 점수: {max_score:.2f}, 낮은 점수: {min_score:.2f}, 총 점수: {total_score:.2f}, 평균 점수: {average_score:.2f}, 분산: {var_score:.2f}')
```

```python
prices = {
    "아메리카노": 2500,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "초코케이크": 5500,
    "샌드위치": 6000
}

def order(**kwargs):
    total = 0

    for menu, count in kwargs.items():

        if menu in prices:
            price = prices[menu] * count
            total += price

            print(f'{menu} {count}개: {price}원')

        else:
            print(f'등록되지 않은 상품입니다: {menu}')

    print(f'총 결제 금액: {total}원')
    print('-' * 20)

order(아메리카노=2, 카페라떼=1)

order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)

order(아메리카노=1, 허니브레드=2)
```
