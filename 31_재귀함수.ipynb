```python
# 재귀호출 사용하기
def hello():
    print('hello,world!')
    hello()

hello()
```

```python
def hello(count):
    if count == 0:
        return 50 # 0이되면 return으로 함수종료
    
    print('hello,word!', count)

    count -= 1
    hello(count)

hello(2)
```

```python
# 재귀호출로 팩토리얼 구하기
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)

print(factorial(5))
```

```python
def fibo(n):
 
    if n == 0:
        return 0
    if n == 1:
        return 1
        
    return fibo(n-2) + fibo(n-1)

n =  int(input())

fibo(n)
```

```python
def recursive_sum(n):
    
    if n == 1:
        return 1
    
    return n + recursive_sum(n-1)

n =  int(input())

print("=" , recursive_sum(n))
```

```python
def recursive_sum(n):

    if n == 1:
        return "1"

    return f"{n} + " + recursive_sum(n - 1)


n = int(input())

total = n * (n + 1) // 2

print(f"{recursive_sum(n)} = {total}")
```
