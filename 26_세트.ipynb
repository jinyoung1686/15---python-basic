```python
# 세트(집합) 만들기
## 세트 = {값1, 값2, 값3}

## 1. 순서가 정해져 있지 않다.
## 2. 중복 될 수 없다.

fruits = {
    'strawberry','grape', 'orange', 'pineapple', 'cherry'
}
print(fruits)
```

```python
fruits[0] # 인덱싱 불가능
```

```python
# 세트에 특정값이 있는지 확인
fruits = {
    'strawberry','grape', 'orange', 'pineapple', 'cherry'
}
print('orange'in fruits)
print('peach'in fruits)
```

```python
# set를 사용하여 세트 만들기
## set(반복가능한객체)

a = set('apple')
print(a)

b = set(range(5))
print(b)

c = set()
print(c)
```

```python
d = {} # 비어있는 중괄호는 dict 
print(type(d))

e = set() # <- 비어있는 중괄호 만들고싶을때
print(type(e))
```

```python
# 집합연산 사용하기
## 세트1 | 세트2
## set.union(세트1, 세트2)
# 합집합
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)
print(set.union(a, b))
print(a.union(b))
```

```python
# 교집합
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a & b)
print(set.intersection(a, b))
print(a.intersection(b))
```

```python
# 차집합
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a - b)
print(set.difference(a, b))
print(a.difference(b))
```

```python
# 대칭 차집합
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a ^ b)
print(set.symmetric_difference(a, b))
print(a.symmetric_difference(b))
```

```python
#집합 연산 후 할당 연산자 사용하기

## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}
a |={5}
print(a)

a.update({6})
print(a)
```

```python
## 세트1 %= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}
print(a)

a = {1, 2, 3, 4}
a.intersection_update({0, 1, 2, 3, 4})
print(a)
```

```python
## 세트1 ^= 세트2
## 세트1.symmetric_difference_update(세트2)

a = {1, 2, 3, 4}
a ^={3, 4, 5, 6}
print(a)

a = {1, 2, 3, 4}
a.symmetric_difference_update({3, 4, 5, 6})
print(a)
```

```python
# 부분집합과 상위집합 확인하기
## 현재세트 <= 다른세트
## 현재세트.issubse(다른세트)

a = {1, 2, 3, 4}
print(a <= {1, 2, 3, 4})
a. issubset({1, 2, 3, 4})
```

```python
## 현재세트 < 다른세트
a = {1, 2, 3, 4}
a < { 1, 2, 3, 4, 5}
```

```python
## 현재세트 >= 다른세트
## 현재세트.issuperset(다른세트)
a = {1, 2, 3, 4}
print(a >= { 1, 2, 3, 4})
a.issuperset({1, 2, 3, 4})
```

```python
## 현재세트 > 다른세트
a = {1, 2, 3, 4}
a > {1, 2, 3}
```

```python
# 세트가 같은지 다른지 확인
a = {1, 2, 3, 4}
print(a == {1, 2, 3, 4})
print(a == {4, 2, 1, 3})
print(a != {1, 2, 3, 4})
print(a != {1, 2, 3})
```

```python
# 세트가 겹치지 않는지 확인학
## 현재세트.isdisjoint(다른세트)
a = {1, 2, 3, 4}
print(a.isdisjoint({5, 6, 7, 8}))
print(a.isdisjoint({3, 4, 5, 6})) # 하나라도 겹치면 False
```

```python
# 세트에 요소 추가하기
a = {1, 2, 3, 4}
a.add(5)
a
```

```python
# 세트에서 특정요소를 삭제하기
a = {1, 2, 3, 4}
a.remove(3)
print(a)

a.discard(2)
print(a)
a.discard(3) # 삭제할 요소가 없으면 그냥 넘어감
print(a)
```

```python
# 세트에서 임의의 요소 삭제하기(앞에서부터삭제됨)
a = {1, 2, 3, 4}
print(a.pop())
print(a)
```

```python
# 반복문으로 세트의 요소를 모두 출력하기

a = {1, 2, 3, 4}
for i in a:
    print(i)
```

```python
# 세트 표현식
## 1.{식 for 변수 in 반복가능한객체}
## 2.set(식 for 변수 in 반복가능객체)

a = {i for i in 'apple'}
print(a)
```

```python
# 세트 표현식에 if 조건문 사용하기
## 1.{식 for 변수 in 세트 if 조건식}
## 2. set(식 for 변수 in 세트 if 조건식)

a = {i for i in 'pineapple' if i not in 'apl'}
print(a)
```

```python
num1, num2 = map(int, input().split())

# 첫 번째 숫자의 약수집합 
a = {i for i in range(1, num1 + 1) if num1 % i == 0}

 # 두 번째 숫자의 약수집합
b = {i for i in range(1, num2 + 2) if num2 % i == 0}

divisor = a & b

result = 0
if type(divisor) == set:
    result = sum(divisor)

print(result)
```
