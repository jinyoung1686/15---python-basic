```python
국어, 영어, 수학, 과학 = map(int,input().split())
total = sum([국어,영어,수학,과학]) 
avg = total / 4
if (국어 < 0 or 국어 > 100 or
    영어 < 0 or 영어 > 100 or
    수학 < 0 or 수학 > 100 or
    과학 < 0 or 과학 > 100):

    print("잘못된 점수")
else:
    if avg >= 80:
        print("합격")
    else:
        print("불합격")
```
